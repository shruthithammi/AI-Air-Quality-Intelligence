```python
air_df = df.copy()
```

```python
missing = air_df.isnull().sum().sort_values(ascending=False)

missing_percentage = (missing / len(air_df) * 100).round(2)

missing_df = pd.DataFrame({
    "Missing Values": missing,
    "Percentage": missing_percentage
})

missing_df
```

The Xylene column contains more than 60% missing values. Since a majority of records are unavailable, this feature is removed to improve data quality and avoid unreliable imputation.

```python
air_df.drop(columns=["Xylene"], inplace=True)
```

```python
air_df = air_df.dropna(subset=["AQI", "AQI_Bucket"])
```

```python
air_df.shape
```

```python
numeric_columns = air_df.select_dtypes(include="number").columns

air_df[numeric_columns] = air_df[numeric_columns].fillna(
    air_df[numeric_columns].median()
)
```

```python
air_df.isnull().sum().sort_values(ascending=False)
```

```python
air_df.to_csv("air_quality_cleaned.csv", index=False)
```

# Data Cleaning

### Steps Performed

1. Converted Date column to datetime format.
2. Checked dataset dimensions and data types.
3. Identified missing values.
4. Removed Xylene column because it contained more than 60% missing values.
5. Removed rows where AQI and AQI_Bucket were missing.
6. Filled remaining missing numerical values using the median.
7. Verified duplicate rows and saved the cleaned dataset.

## EXPLORATORY DATA ANALYSIS

## 1. Distribution of Air Quality Index (AQI)


```python
import matplotlib.pyplot as plt
import plotly.express as px
import seaborn as sns
```

```python
fig = px.histogram(
    air_df,
    x="AQI",
    nbins=40,
    title="Distribution of Air Quality Index (AQI)",
    color_discrete_sequence=["#2E8B57"]
)

fig.show()
```

### Observation

* The AQI values are heavily concentrated in the lower range, while only a small number of records have very high AQI values.
* The distribution is right-skewed, which indicates that extremely polluted days occur less frequently than normal or moderate pollution days.
* This suggests the dataset contains both regular air quality conditions and a few severe pollution events that should be handled carefully during model training.


## 2. Distribution of AQI Categories


```python
fig = px.histogram(
    air_df,
    x="AQI_Bucket",
    color="AQI_Bucket",
    title="Distribution of AQI Categories"
)

fig.show()
```

### Observation

* The **Moderate** and **Satisfactory** categories appear most frequently in the dataset.
* **Good** and **Severe** air quality categories have comparatively fewer observations.
* Since the dataset includes multiple AQI categories, it can also be used for an AQI classification problem in addition to AQI prediction.


## 3. Top 10 Most Polluted Cities (Average AQI)


```python
city_pollution = (
    air_df.groupby("City")["AQI"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

fig = px.bar(
    city_pollution,
    x="City",
    y="AQI",
    color="AQI",
    title="Top 10 Most Polluted Cities (Average AQI)"
)

fig.show()
```

### Observation

* Ahmedabad has the highest average AQI among the cities in this dataset, followed by Delhi and Patna.
* Most cities in this list are large urban or industrial regions where pollution levels remain consistently high.
* This visualization helps identify cities that experience poor air quality more frequently over the observed period.

## 4. Top 10 Least Polluted Cities (Average AQI)


```python
least_pollution = (
    air_df.groupby("City")["AQI"]
    .mean()
    .sort_values()
    .head(10)
    .reset_index()
)

fig = px.bar(
    least_pollution,
    x="City",
    y="AQI",
    color="AQI",
    title="Top 10 Least Polluted Cities"
)

fig.show()
```

### Observation

* Aizawl and Shillong have the lowest average AQI values in the dataset, indicating comparatively cleaner air quality.
* Many cities in this list are less industrialized or have favorable geographical conditions that contribute to better air quality.
* Comparing this graph with the most polluted cities provides a clear understanding of regional differences in pollution levels.

## 5. Correlation Between Pollutants and AQI

```python
plt.figure(figsize=(12,8))

corr = air_df.select_dtypes(include="number").corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Air Pollutants")
plt.show()
```

### Observation

* PM2.5 has a strong positive correlation with AQI (**0.66**), which indicates that higher PM2.5 levels are associated with poorer air quality.
* Carbon Monoxide (CO) also shows a strong positive relationship with AQI (**0.68**), followed by NO₂ (**0.54**) and SO₂ (**0.49**).
* PM10 has a moderate correlation with AQI (**0.46**), while pollutants like Benzene and O₃ have relatively weak correlations.
* NO and NOx are highly correlated with each other (**0.79**), suggesting that these pollutants often increase together.
* Overall, PM2.5, CO, NO₂, PM10, and NOx appear to be the most informative features for predicting AQI.

## 8. Monthly Average AQI Trend


```python
air_df["Month"] = air_df["Date"].dt.month_name()

month_order = [
    "January","February","March","April","May","June",
    "July","August","September","October","November","December"
]

monthly_aqi = (
    air_df.groupby("Month")["AQI"]
    .mean()
    .reindex(month_order)
    .reset_index()
)

fig = px.line(
    monthly_aqi,
    x="Month",
    y="AQI",
    markers=True,
    title="Monthly Average AQI Across India"
)

fig.show()
```

### Observation

* AQI is highest during **October, November, December, and January**, indicating poorer air quality during winter months.
* The average AQI gradually decreases from February and reaches its lowest levels during the monsoon period around July and August.
* This seasonal trend suggests that weather conditions play an important role in air pollution levels.


## 9. Year-wise Average AQI Trend


```python
air_df["Year"] = air_df["Date"].dt.year

yearly_aqi = (
    air_df.groupby("Year")["AQI"]
    .mean()
    .reset_index()
)

fig = px.line(
    yearly_aqi,
    x="Year",
    y="AQI",
    markers=True,
    title="Year-wise Average AQI (2015–2020)"
)

fig.show()
```

### Observation

* The average AQI shows an overall decreasing trend from 2015 to 2020.
* A noticeable drop is observed in 2020 compared to previous years.
* This indicates that air quality improved during the later years covered by the dataset, although additional factors would be needed to explain the reason behind this trend.

## 10. Hyderabad Air Quality Trend


```python
# Hyderabad AQI Trend (Fixed for latest Pandas)

hyderabad = air_df[air_df["City"] == "Hyderabad"]

monthly_hyd = (
    hyderabad
    .resample("ME", on="Date")["AQI"]   # Changed M → ME
    .mean()
    .reset_index()
)

fig = px.line(
    monthly_hyd,
    x="Date",
    y="AQI",
    markers=True,
    title="Monthly AQI Trend in Hyderabad"
)

fig.show()
```

### Observation

* Hyderabad's AQI changes throughout the year instead of remaining constant, showing clear seasonal fluctuations.
* Higher AQI values are observed during the winter months, while AQI decreases during the monsoon period, indicating comparatively cleaner air.
* A noticeable decline in AQI is also visible towards 2020, suggesting an improvement in air quality during the later part of the dataset.
* This trend shows that both season and time can influence AQI levels, making them useful features for machine learning models.
